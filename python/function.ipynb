Types of Functions in Python

Functions in Python are mainly of two types:

1. Built-in Functions

These are predefined functions already available in Python.

Examples
print() → Displays output
len() → Finds length
input() → Takes input from user
type() → Shows data type
Example Program
name = "Python"
print(len(name))
Output
6
2. User-defined Functions

These are functions created by the programmer using the def keyword.

Syntax
def function_name():
    # code
Example
def greet():
    print("Welcome to Python")

greet()
Output
Welcome to Python
Other Common Types of User-defined Functions
a) Function Without Arguments and Without Return Value
def show():
    print("Hello")

show()
b) Function With Arguments
def add(a, b):
    print(a + b)

add(2, 3)
c) Function With Return Value
def square(n):
    return n * n

print(square(5))
d) Recursive Function

A function that calls itself.

def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print(fact(5))
Output
120
e) Lambda Function

A small anonymous function written in one line.

square = lambda x: x * x
print(square(4))
Output
16

```python
def grade(p):
    print(p)
    for x in p:
        if x<= 100 and x >=90:
            print("A++")
        elif x<= 90 and x>=80:
            print("A")
        elif x<=70 and x >= 60:
            print("B")
        elif x<=50 and x >= 40:
            print("C")
        else:
            print("D")
p1 = [34,50,40,90,22,4]
grade(p = p1)
```
