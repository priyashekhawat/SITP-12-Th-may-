```python
sns.lineplot(df["owner"])
plt.show()
```

```python
plt.plot(df['brand'])
plt.show()
```

```python
sns.histplot(df['brand'], kde=True)
plt.title("Bike Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.xticks(rotation=90)
plt.show()
```

```python
sns.countplot(x='brand', data=df)
plt.title("Number of Bikes by Brand")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```

```python
sns.kdeplot(df['price'])
plt.show()
```

```python
sns.countplot(x='owner', data=df)
plt.title("Number of Bikes by Brand")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.xticks(rotation=360)
plt.show()
```

```python
sns.scatterplot(
    x='kms_driven',
    y='price',
    hue='brand',
    data=df
)

plt.title("Price vs Age")
plt.xlabel("km_driven")
plt.ylabel("Price")
plt.show()
```

```python
sns.catplot(x= 'owner',y = 'price',kind = 'bar',data= df)
plt.title("price distribution of owner")
plt.xticks(rotation=60)
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='brand',
    y='price',
    data=df
)

plt.title("Price Distribution by Brand")
plt.xlabel("Brand")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.violinplot(
    x='owner',y='price',data=df)

plt.title("Price Distribution by Owner Type")
plt.xlabel("Owner Type")
plt.ylabel("Price")
plt.show()
```

```python
brand_price = df.groupby('brand')['price'].mean()

brand_price.plot(kind='bar')
plt.title("Average Bike Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Price")
plt.show()
```

```python
brand_count = df['owner'].value_counts()

plt.pie(
    brand_count.values,
    labels=brand_count.index,
    explode = [0,0,0,1],
    shadow= True,
    autopct='%1.1f%%'
)

plt.title("Brand Distribution")
plt.show()
```

```python
price_sorted = df['price'].sort_values()

plt.plot(price_sorted.values)
plt.title("Sorted Bike Prices")
plt.xlabel("Bike Index")
plt.ylabel("Price")
plt.show()
```

```python
sns.relplot(x='age', y='price', hue='brand',data=df)
plt.show()
```
