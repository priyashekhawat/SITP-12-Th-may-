## Inheritance
Inheritance is one of the most important concepts of Object-Oriented Programming (OOP).
It allows one class to use the properties and methods of another class.

Inheritance helps programmers:
Reuse code
Save time
Reduce errors
Make programs organized


### pillars of Object-Oriented Programming (OOP).
|Parent Class|	The class whose properties are inherited|
|--------------|----------------------------------------------------|
|Child Class|The class that inherits properties from another class|
|Base Class|	Another name for parent class.|
|Derived Class|	Another name for child class.|
|Main Purpose|	Code reusability and reducing duplication.|

#### Benefits
	Reuse code
    save time
    easy maintenance
    organized code.





### Types of Inheritance
|Type of Inheritance|	Description|
|-------------------|-------------------------------------------------|
| Single Inheritance|	One child class inherits one parent class|
|Multiple Inheritance|	One child class inherits multiple parent classes|
|Multilevel Inheritance|	A class inherits from another derived class|
|Hierarchical Inheritance|	Multiple child classes inherit one parent class| 
|Hybrid Inheritance|	Combination of two or more inheritance types|

```python
# single inheritance

class College :
    college_name = "SGI"
    college_address = "siker"
    collegesp = 46  # ek class se method ke ander properties inherit karne ke liye self is used
    def show(self):
        print(self.college_name)
        print(self.college_address)
        print(self.collegesp)

class student(College):
    student_name = "priya"
    student_rollnumber = 39
    student_number = 89046768
    student_branch = "cse"
st = student()
st.show()
print(st.student_branch)




```

```python
# multilevel inheritance

class Person:
    pass
class Teacher(Person):
    pass
class HOD(Teacher):
    pass
```

```python
class Person:
    def person_info(self):
        print("i am person")
class Teacher(Person):
    def teacher_info(self):
        print("i am teacher")
class HOD(Teacher):
    def hod_info(self):
        print(" i am hod")


h = HOD()
print(h.person_info())
print(h.teacher_info())
print(h.hod_info())
```

```python
# mutiple inheritance

class Developer:
    pass
class Manager:
    pass
class teamleader(Developer,Manager):
    pass

```

```python
class Developer:
    def codding(self):
        print("yeh code write karega")

class Manager:
    def meeting(self):
        print("isko kuch nhi aata isha ek hi kam hai meeting ")

class Teamleader(Developer,Manager):
    def project(self):
        print("project ko handle karna")
team = Teamleader()
team.codding()
```

```python
# create company an excess two company data excess

class Company1:
    def company1_data(self):
        print(" data of the first company")

class Company2:
    def company2_data(self):
        print("data of second company")

class Company3(Company1,Company2):
    def company3_data(self):
        print("data of third company")

company = Company3()
company.company3_data()
company.company1_data()
company.company2_data()
```

```python
### Heiarical
class Developer:
    def codding(self):
        print("yeh code write karega")

class Manager(Developer):
    def meeting(self):
        print("isko kuch nhi aata isha ek hi kam hai meeting ")

class Teamleader(Developer):
    def project(self):
        print("project ko handle karna")

class Designer(Developer):
    print("i am a designer")

team = Teamleader()
man = Manager()
des = Designer()

man.codding()
team.codding()
des.codding()

```

```python
## hybrid

class son:
    def son(a):
        print(" i am a son")
class daughter:
    def daughter(b):
        print("i am  daughter")
class mother(son,daughter):
    def mother(c):
        print("i am a mother")
class father(mother):
    def father(self):
        print(" i am a father")

mom = mother()
dad = father()
mom.daughter()
dad.son()       
```
