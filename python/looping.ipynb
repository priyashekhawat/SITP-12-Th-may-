## Looping in python 
*A loop is used to execute a block of code repeatedly until a condition becomes false.

### Python mainly provides two types of loops:

* for loop

* while loop

 <span style="color:orange">1) For loop </span>

A for loop is used to iterate over a sequence such as a list, tuple, string, or range.

<span style="color:orange"> 2) While loop</span>

A while loop executes as long as the condition is true.

```python
name ="priya"
for x in name :
    print(x)
```

```python
lst = [4,8,9,0]
for y in lst :
    print(y)
```

```python
for x in range(1,20):
    print(x)
```

```python
for x in range(2,22,2):
    print(x)
```

```python
dict = {1:"priya",
        2:"garima",
        3:"swati"}
print(dict)
for x in dict.items():  # same for keys and values
    print(x)
```

```python
tpl = (1,2,3,2,5,2,0,5,5,8)
for item in tpl:
    print(item)
```

```python
lst = [1,2,3,4,5]
for x in lst:
    if x == 3:
        break # it is used to break the loop from passed element
    print(x)
```

```python
lst = [1,2,3,4,5]
for x in lst:
    if x == 3:
        continue  # it is used to skip the element which we passed
    print(x)
```

```python
fruits = ["apple","orange","pinapple","mango"]
for x in fruits:
    if x == "mango":
        continue
    print(x)
```

```python
lst = [1,2,56,89,["hello","hii","jow"]]
print(lst[4])
print(lst[4][1])
```

## while loop

```python
x = 1
while x<=5:
    print(x)
    x+=1
```

```python
x = 5
while x >=1:
    print(x)
    x-=1
```

```python
x = 1
while x<=10:
    if x == 5:
        break
    print(x)
    x+=1
```
