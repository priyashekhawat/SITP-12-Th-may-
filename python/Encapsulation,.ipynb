#### Encapulation

```python
class college:
    college_name = "sgi"  ## all  three are public variable
    college_garden = "yes"
    college_account = 7864566
```

```python
clg = college()
```

```python
clg.college_account
```

```python
class college:
    college_name = "sgi"   # public variable 
    college_garden = "yes" # public variable
    __college_account = 7864566  # private variable (by double underscore)
clg = college()
```

```python
#clg.__college_account   ## it show error because it is private variable
```

```python
class college:
    college_name = "sgi"  ## public variable
    _college_garden = "yes"  ## protected  (by single underscore)
    college_account = 7864566  ## public

clg = college()
```

```python
clg.college_garden
```

```python
class college:
    college_name = "sgi" 
    college_garden = "yes"  
    __college_account = 7864566  
    def display(self):   ## self is object , keyword, and parameter
        return self.__college_account   
clg = college()
clg.display()
```

```python
class college:
    college_name = "sgi" 
    college_garden = "yes"  
    __college_account = 7864566  
    def display(self,a,b):  
        print(a + b)
clg = college()
clg.display(a = 10,b = 20)
```
