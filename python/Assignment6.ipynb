Oues-1) Create a class Employee with attributes emp_name, emp_id, and salary. Create 3 objects and display their details

```python
class Employee:
    def __init__(self, emp_name, emp_id, salary):
        self.emp_name = emp_name
        self.emp_id = emp_id
        self.salary = salary

    def display(self):
        print("Emloyee name:-",self.emp_name)
        print("Employee id:-",self.emp_id)
        print("employee salary:-",self.salary)

# creating 3 objects:
e1 = Employee("Priya",1,50000)
e2 = Employee("Garima",2,70000)
e3 = Employee("Swati",3,40000)

# displaying the details:
e1.display()
e2.display()
e3.display()





```

Oues-2) Create a class Calculator with methods for addition, subtraction, multiplication, and division

```python
class Calculator:
    def add(self,a,b):
        print("Addition:-",(a+b))
    def subtract(self,c,d):
        print("Subtration:-",(c-d))
    def multiply(self,e,f):
        print("Multiplication:-",(e*f))
    def division(self,g,h):
        print("Division:-",(g/h))
    
cal = Calculator()
cal.add(9,5)
cal.subtract(9,5)
cal.multiply(9,5)
cal.division(9,5)

    
```

Oues-3) Create a class Book with a constructor to initialize title, author, and price, then print details.

```python
class Book:
    def __init__(self,title,author,price):
        self.title = title
        self.author = author
        self.price = price
    def display(self):
        print("Book title:-",self.title)
        print("Book author:-",self.author)
        print("Book price:-",self.price)
    
b1 = Book("The unpaved path","Arvind",500)

b1.display()
```
