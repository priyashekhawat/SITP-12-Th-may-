### File handling
File Modes
|Mode|	Meaning|
|------|-----------|
|"r"|	Read mode|
|"w"|	Write mode|
|"a"	|Append mode|
|"x"	|Create new file|
|"b"|	Binary mode|
|"t"	|Text mode|

### Important File Functions
|Function|	Work|

read()|	Reads full file
readline()	Reads one line
readlines()	Reads all lines as list
write()	Writes data
close()	Closes file

```python
file=open("poit.txt",'r')
```

```python
file.read()
```

```python
file.readline()
```

```python
file.readlines()
```

```python
file.close()
```

```python
with open('poit.txt') as file:
    print(file.read())
    # we can open file by this method too
```

```python
file= open('poit.txt','w')  # we can use w for create the new file like we can write poit1 and file was created
file.write("hii how are you  ")  # write is used to change the text
file.close()
file =open('poit.txt','r')
file.read()
```

```python
file = open('poit.txt','a')
file.write("File handling means creating, opening, reading, writing, and closing files using a program.")
file.close()
```

```python
open('poit.txt','r').read()
```

```python
file = open('filehandling.txt','a')   # this will create a new file 
file.write("File handling means creating, opening, reading, writing, and closing files using a program.")
file.close()
```
