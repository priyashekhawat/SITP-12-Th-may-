```python
class InsufficientBalance(Exception):  # we can take class as any error 
    pass

try:
    bal = 3000
    a = int(input("enter your withdrawal amount"))
    if a > bal :
        raise InsufficientBalance ("not have sufficient balance")
    print("you can take money")
except InsufficientBalance as e:
    print(e)
```

```python
class PasswordLen(Exception):
    pass
try:
    password = 1234567
    a = int(input("enter your password"))
    if  a!= password:
        raise PasswordLen ("length of password is not correct")
    print("correct password")
except PasswordLen as e:
    print(e)
```

```python
class Email_check(Exception):
    pass
try:
    email = input("enter your email")
    if "@" not in email:
        raise Email_check("email have not @ :-",email)
    if "." not in email:
        raise Email_check("email not have .:-",email)
    if isupper() not in email:
        raise Email_check("email must contain one uppercase:-",email)
    if islower() not in email:
        raise Email_check("email must contain one lowercase:-",email)
    print("correct email:-",email)
except Email_check as e:
    print(e)
```
