```python
xpoint = [1,2,3,4]
ypoint = [2,4,6,8]
plt.plot(xpoint,ypoint, ms = 20 , marker = 'o')
plt.show()
# ms shows size of marker 
```

```python
xpoint = [1,2,3,4]
ypoint = [2,4,6,8]
plt.plot(xpoint,ypoint, ms = 20 , marker = '*')  # not use other special charter 
plt.show()
```

```python
xpoint = [1,2,3,4]
ypoint = [2,4,6,8]
plt.plot(xpoint,ypoint, ms = 20 , marker = '*', mfc = 'black', mec = "red")
plt.show()
# mfc is used to colour the point and mec is used to colour the outline
```

```python
# pie chart
a = [1,2,3,4]
plt.pie(a)
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","noddles","fries"]
plt.pie(a, labels = label)
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","noddles","fries"]
plt.pie(a, labels = label, explode = [0,0.2,0,0])
plt.show()
# explode is used to out the item from graph and take all other 0.
```

```python
a = [1,2,3,4]
label = ["burger","pizza","noddles","fries"]
plt.pie(a, labels = label, explode = [0,0.2,0,0], shadow = True)
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","noddles","fries"]
c = ["pink","brown","yellow","grey"]
plt.pie(a, labels = label, explode = [0,0.2,0,0], shadow = True,colors = c)
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","noddles","fries"]
c = ["pink","brown","yellow","grey"]
plt.pie(a, labels = label, explode = [0,0.2,0,0],shadow = True, colors = c, autopct = '%1.1f%%')
plt.show()
```

```python
a = [1,2,3,4]
label = ["burger","pizza","noddles","fries"]
c = ["pink","brown","yellow","grey"]
plt.pie(a, labels = label, explode = [0,0.2,0,0], colors = c,shadow = True,autopct = '%1.1f%%')
plt.legend()
plt.show()
# legend show the status of the pie chart
```
